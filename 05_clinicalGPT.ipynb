```python
from transformers import AutoTokenizer, AutoModelForCausalLM
tokenizer = AutoTokenizer.from_pretrained("medicalai/ClinicalGPT-base-zh")
model = AutoModelForCausalLM.from_pretrained("medicalai/ClinicalGPT-base-zh")
```

```python
question = "What are the symptoms of COVID-19?"

# Encode the question
encoded_question = tokenizer.encode_plus(question, return_tensors='pt')

# Pass the encoded question to the model
outputs = model(**encoded_question)
```

```python
question = "What could be the diagnosis for a black person aged 43 with bmi 30 smoker low wbc count and high platelet count with flu-like symptoms and a body temperature of 38 degree Celsius"
```

```python
encoded_question = tokenizer.encode_plus(question, return_tensors='pt')
outputs = model(**encoded_question)
```

```python
token_ids = torch.argmax(outputs.logits, dim=-1)
for i in range(len(token_ids)):
    answer = tokenizer.decode(token_ids[i])
    print(answer)
```

```python
sentence = 'Hello, doctor' 
batch = tokenizer(
            sentence,
            return_tensors="pt", 
            add_special_tokens=False
        )
```

```python
generated = model.generate(inputs = batch["input_ids"], max_length=200, do_sample=True, top_k=50)
print("model predict: ", tokenizer.decode(generated[0]))
```
