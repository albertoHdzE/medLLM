```python
import json
def merge_response(response) -> str:
    out = ''
    list_dictionaries = response.text.split('data: ')
    # print('******')
    # print(list_dictionaries)
    # print('******')
    for d in list_dictionaries:
        if len(d) > 0:
            # print('line: ', d)
            one_dict = json.loads(d)
            out += one_dict['content']
    out = out.replace('\n', '')
    return out
```

```python
import os

# Import writer class from csv module
from csv import writer
def append_result_csv(results_list: list, out_file_name: str):
    # Open our existing CSV file in append mode
    # Create a file object for this file
    if not os.path.isfile(out_file_name):
        with open(out_file_name, 'a') as f_object:
            writer_object = writer(f_object)
            # writer_object.writerow(["complexity", "length_type", "language", "q_length", "a_length", "temperature", "print_code", "real_code", "incapacity"])
            writer_object.writerow(["complexity", "length_type", "language", "q_length", "a_length", "temperature"])
            writer_object.writerow(results_list)
    else:
        with open(out_file_name, 'a') as f_object:
            # Pass this file object to csv.writer()
            # and get a writer object
            writer_object = writer(f_object)
            # Pass the list as an argument into
            # the writerow()
            writer_object.writerow(results_list)
            # Close the file object
            f_object.close()


def append_answers_log(line, out_file_name):
    if not os.path.isfile(out_file_name):
        with open(out_file_name, 'a') as f_object:
            writer_object = writer(f_object)
            f_object.write(line + "," + "\n")
            f_object.close()
    else:
        with open(out_file_name, 'a') as f_object:
            writer_object = writer(f_object)
            f_object.write(line + "," + "\n")
            f_object.close()
```

```python
def extract_by_temperature_by_language(file_path, temperature_value, language=""):
    df = pd.read_csv(file_path)
    mal1 = pd.DataFrame()
    mal2 = pd.DataFrame()
    mal3 = pd.DataFrame()
    pyl1 = pd.DataFrame()
    pyl2 = pd.DataFrame()
    pyl3 = pd.DataFrame()
    mpl1 = pd.DataFrame()
    mpl2 = pd.DataFrame()
    mpl3 = pd.DataFrame()

    # only gets real_code and explanationss, ignores prints, no code and non_senses
    #df = df[((df["real_code"] == 1) | (df["explanation"] == 1)) & (df["no_code"] == 1)]
    df = df[((df["real_code"] == 1) | (df["explanation"] == 1))]
    # print("********************")
    # print(df["no_code"].unique())
    # print(df["print_code"].unique())
    # print("********************")

    if language == "Mathematica" or language == "":
        mal1 = df[(df["language"]=="Mathematica") & (df["complexity"]=="1") & (df["temperature"]==temperature_value)]
        mal2 = df[(df["language"]=="Mathematica") & (df["complexity"]=="2") & (df["temperature"]==temperature_value)]
        mal3 = df[(df["language"]=="Mathematica") & (df["complexity"]=="3") & (df["temperature"]==temperature_value)]

    if language == "Python" or language == "":
        # print("im here *******")
        pyl1 = df[(df["language"]=="Python") & (df["complexity"]=="1") & (df["temperature"]==temperature_value)]
        pyl2 = df[(df["language"]=="Python") & (df["complexity"]=="2") & (df["temperature"]==temperature_value)]
        pyl3 = df[(df["language"]=="Python") & (df["complexity"]=="3") & (df["temperature"]==temperature_value)]
        # print(pyl1)
        # print(pyl2)
        # print(pyl3)
    if language == "Maple" or language == "":
        mpl1 = df[(df["language"]=="Maple") & (df["complexity"]=="1") & (df["temperature"]==temperature_value)]
        mpl2 = df[(df["language"]=="Maple") & (df["complexity"]=="2") & (df["temperature"]==temperature_value)]
        mpl3 = df[(df["language"]=="Maple") & (df["complexity"]=="3") & (df["temperature"]==temperature_value)]

    indexes = [1, 2, 3] # three levels of complexity
    new_df = pd.DataFrame()
    new_df["index"] = indexes
    if len(mal1) > 0:
        new_df["math_avgs_"+str(temperature_value)] = [mal1["a_length"].mean(), mal2["a_length"].mean(), mal3["a_length"].mean()]
    if len(pyl1) > 0:
        new_df["py_avgs_"+str(temperature_value)] = [pyl1["a_length"].mean(), pyl2["a_length"].mean(), pyl3["a_length"].mean()]
    if len(mpl1) > 0:
        new_df["mple_avgs_"+str(temperature_value)] = [mpl1["a_length"].mean(), mpl2["a_length"].mean(), mpl3["a_length"].mean()]

    return new_df
```

```python
def return_indexes_by_filter(file_path,
                            valid=0,
                            no_code=0,
                            print_code=0,
                            real_code=0,
                            explanation=0,
                            no_sense=0,
                            language="Python",
                            temperature=-1, # -1 means "all teperatures"
                            complexity="0", # [1-3]. -1 means "all complexities"
                            length_type=-1, # [0-2], -1 menas "all lengths"
                            verbose=0
                            ):
    
    df = pd.read_csv(file_path)

    filters = {"no_code": no_code,
            "print_code": print_code,
            "real_code": real_code,
            "explanation": explanation,
            "no_sense": no_sense}
    
    if temperature > 0:
        filters["temperature"] = temperature
    if complexity != "0":
        filters["complexity"] = "\'" + complexity + "\'"
    if length_type > -1:
        filters["length_type"] = length_type
        
    query = ' & '.join(['{}=={}'.format(k, v) for k, v in filters.items()])
    if verbose == 1:
        print("Filters: ")
        print(filters)
        print("QUERY: ", query)

    filtered_df = df.query(query)

    if verbose==1:
        print("found instances: ")
        print(filtered_df)
    indexes = filtered_df.index.to_list()
    return indexes, filtered_df
```

```python
def get_answers_by_filter(file_path,
                            no_code=0,
                            print_code=0,
                            real_code=0,
                            explanation=0,
                            no_sense=0,
                            language="Python",
                            temperature=-1, # -1 means "all teperatures"
                            complexity="0", # [1-3]. -1 means "all complexities"
                            length_type=-1, # [0-2], -1 menas "all lengths"
                            verbose=0):
    
    indexes, filtered_df = return_indexes_by_filter(file_path=file_path,
                                        no_code=no_code,
                                        print_code=print_code,
                                        real_code=real_code,
                                        explanation=explanation,
                                        no_sense=no_sense,
                                        language=language,
                                        temperature=temperature,
                                        complexity=complexity,
                                        length_type=length_type,
                                        verbose=verbose)
    print("Indexes:")
    print(indexes)
    
    results = []
    for i in indexes:
        line = read_line_by_index(file_path.replace(".csv", "_log.csv"), i)
        results.append(line)
        if verbose == 1:
            df = pd.read_csv(file_path)
            print(i,': ',df.iloc[[i]].values)
            print(read_line_by_index(file_path.replace(".csv", "_log.csv"), i))
    return results
    
```

```python
def recalculate_avg_valid_answers(file_path, temperature, complexity, language="Python", verbose=0):
    real_code_lines = get_answers_by_filter(file_path,
                                            real_code=1,
                                            complexity=complexity,
                                            temperature=temperature,
                                            language=language,
                                            verbose=verbose)
    
    print("real code lines")
    print(real_code_lines)
    explanation_lines = get_answers_by_filter(file_path,
                                            explanation=1,
                                            complexity=complexity,
                                            temperature=temperature,
                                            language=language,
                                            verbose=verbose)
    
    all_lines = []
    for l in real_code_lines:
        all_lines.append(l)
    for l in explanation_lines:
        all_lines.append(l)
    counter_lines = 0
    total_len = 0
    for l in all_lines:
        counter_lines +=1
        total_len = total_len + len(l)
        
    return total_len/counter_lines
    
```
