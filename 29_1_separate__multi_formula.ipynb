```python
import pandas as pd
```

```python
formulas_ds = pd.read_csv('new-multi-formulae.csv')
formulas_ds
```

```python
model_names = [col for col in formulas_ds.columns if col not in {'sequence', 'Complexity'}]
model_names
```

```python
import ast
import json

def separate_formulas(df, model_column):
    # Find the maximum number of formulas in the column
    max_formulas = 0
    
    for index, row in df.iterrows():
        try:
            sample = row[model_column]
            print('processing: ', sample)
            sample_list = json.loads(sample.replace("'", '"'))
            # Only count non "*not found" entries for max_formulas
            valid_formulas = [f for f in sample_list if f != "*not found"]
            max_formulas = max(max_formulas, len(valid_formulas))
        except Exception as e:
            print(f"Error processing row {sample}, {index}: {e}")
    
    # Create new columns based on max_formulas
    for i in range(max_formulas):
        column_name = f"{model_column}_{i+1}"
        df[column_name] = '*not found'
    
    # Populate the new columns with formulas
    for index, row in df.iterrows():
        try:
            sample = row[model_column]
            sample_list = json.loads(sample.replace("'", '"'))
            
            # Filter out "*not found" entries
            valid_formulas = [f for f in sample_list if f != "*not found"]
            
            # Assign each valid formula to its corresponding column
            for i, formula in enumerate(valid_formulas):
                column_name = f"{model_column}_{i+1}"
                df.at[index, column_name] = formula
        except Exception as e:
            print(f"Error processing row {index}: {e}")
    
    return df

# Usage example: apply to all model names
for model in model_names:
    formulas_ds = separate_formulas(formulas_ds, model)
formulas_ds.tail()
```
