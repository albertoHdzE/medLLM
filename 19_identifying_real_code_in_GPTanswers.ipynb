# MATLAB

````python
import pandas as pd
import re
````

````python
matlab_df = pd.read_csv("Matlab_evaluation.csv")
matlab_df
````

````python
answers = list(matlab_df["GPT4_answer"].values)
# answers_test = answers[450:590]
# answers_test
answers
````

````python
def preprocess_code(code):
    # Remove leading and trailing substrings
    code = re.sub(r'^```(?:Matlab|matlab)?\n|\n```$', '', code)
    return code

def find_matched_programs(matlab_programs):
    # Regular expression pattern
    pattern = r'^(?!.*(?:\bfor\b|\bwhile\b|\bdo\b|\brepeat\b))(?!.*[-+*/^:])(?!.*\bprime\b).*(?:\bdisp\s*\(\s*(?:\w+\s*=\s*\[?\d+(?:,\s*\d+)*\]?|\[\s*\d+(?:,\s*\d+)*\s*\]|\d+(?:,\s*\d+)*)\s*\)|\w+\s*=\s*\[?\d+(?:,\s*\d+)*\]?;\s*disp\(\w+\)\s*)\s*;?$'
    
    # List to store matched programs and their indices
    matched_programs = []
    indexes = []

    # Iterate over each MATLAB code string
    for i, program in enumerate(matlab_programs):
        # Preprocess the code string
        program = preprocess_code(program)
        # Check if the program matches the pattern
        if re.match(pattern, program):
            matched_programs.append(program)
            indexes.append(i)
    
    return matched_programs, indexes
````

````python
# Find MATLAB programs containing sequences of numbers and disp instructions
matches, indexes= find_matched_programs(answers)

# Print matching MATLAB programs and their corresponding index
print("Matching MATLAB programs:")
for i in range(len(matches)):
    print(f"Index: {indexes[i]}, Code: {matches[i]}")
````

````python
indexes
````

````python
matlab_df.loc[indexes, 'print_code'] = 1
matlab_df
````

````python
matlab_df.to_csv("automatic_matlab.csv", index=False)
````

# Python

````python
python_df = pd.read_csv("python_evaluation.csv")
answers = list(python_df["GPT4_answer"].values)
answers
````

````python
def preprocess_code_p(code):
    # Remove leading and trailing substrings
    code = re.sub(r'^```(?:Python|python)?\n|\n```$', '', code)
    return code

def find_matched_programs_p(python_programs):
    # Regular expressions for each rule
    patterns = [
        # Rule: Declaration of a variable as a list of numbers and then print of such variable
        r'^(?!.*(?:for|while|import|def))(?!.*[-+*/^:])(?!.*:).*\bsequence\s*=\s*\[\s*\d+(?:,\s*\d+)*\s*\]\s*;?\s*\bprint\s*\(\s*sequence\s*\)\s*$',
        # Rule: Declaration of a variable as a list or sequence of numbers and then print of such variable
        r'^(?!.*(?:for|while|import|def))(?!.*[-+*/^:])(?!.*:).*\bprint\s*\(\s*\w+\s*=\s*\[\s*\d+(?:,\s*\d+)*\s*\]\s*\)\s*$',
        # Rule: List of variables to represent a sequence of numbers and then print such list of variables
        r'^(?!.*(?:for|while|import|def))(?!.*[-+*/^:])(?!.*:).*\bprint\s*\(\s*(?:\w+(?:\s*,\s*\w+)*)\s*\)\s*$',
        # Rule: Program that prints a list or sequence of numbers
        r'^(?!.*(?:for|while|import|def))(?!.*[-+*/^:])(?!.*:).*\bprint\s*\(\s*(?:\[\s*\d+(?:,\s*\d+)*\s*\]|\d+(?:,\s*\d+)*)\s*\)\s*$',
    ]
    
    # List to store matched programs and their indices
    matched_programs = []
    indexes = []

    # Iterate over each Python code string
    for i, program in enumerate(python_programs):
        # Preprocess the code string
        program = preprocess_code_p(program)
        # Flag to track if program has been matched by any rule
        matched = False
        # Iterate over each rule
        for pattern in patterns:
            # Check if the program matches the pattern
            if re.match(pattern, program, re.MULTILINE):
                matched_programs.append(program)
                indexes.append(i)
                matched = True
                break  # Move to the next program if a match is found
        # If no match is found, print a message
        if not matched:
            print(f"No match found for program at index {i}")
    
    return matched_programs, indexes
````

````python
matched, indexes = find_matched_programs_p(answers)

for i in range(len(matched)):
    print(f"Match found at index {indexes[i]}:")
    print(matched[i])
````

````python
python_df.loc[indexes, 'print_code'] = 1
python_df
````

````python
python_df.to_csv("automatic_python.csv", index=False)
````

# Mathematica

````python
mathe_df = pd.read_csv("mathematica_evaluation.csv")
answers = list(mathe_df["GPT4_answer"].values)
answers
````

````python
test_answers = [
    '```mathematica\nFindSequenceFunction[{62, 98, 130, 154, 290, 315, 324, 385, 408, 447}, n]\n```', # no
    '```mathematica\nFindSequenceFunction[{2, 42, 66, 102, 153, 195, 201, 252, 306, 396}, n]\n```', # no
    '```mathematica\nSequence[187, 218, 260, 295, 301, 314, 379, 410, 452, 469]\n```', # yes
    '```mathematica\nSelect[Range[1000], PrimeQ[# + 1] && PrimeQ[# + 3] &]\n```', # no
    '```mathematica\na[n_] := If[n == 1, 4, If[EvenQ[n], a[n - 1] + 7*n, a[n - 1] + 18*(n - 1)]];\nTable[a[n], {n, 1, 10}]\n```', # no
    '```mathematica\nSelect[Range[100, 1000], Max[IntegerDigits[#]] == 5 &]\n```', # no
    '```mathematica\nSequence[29, 63, 95, 140, 150, 190, 221, 437, 482, 491]\n```', # yes
    '```mathematica\nSequence[72, 106, 139, 165, 171, 192, 199, 429, 453, 477]\n```', # yes
    '```mathematica\nTable[n + Sum[If[Mod[n, i] == 0, i, 0], {i, 1, n - 1}], {n, 15, 450}]\n```', #no 
]
````

````python
import re

def preprocess_code_mathematica(code):
    # Remove leading and trailing substrings
    code = re.sub(r'^```(?:Mathematica|mathematica)?\n|\n```$', '', code)
    return code

def find_matched_programs_mathematica(mathematica_programs):
    # Regular expression patterns to match different cases
    patterns = [
        r'^\s*Print\s*\[\s*(?:\{\s*\d+(?:,\s*\d+)*\s*\}|\d+(?:,\s*\d+)*)\s*\]\s*;',
        r'^\s*Sequence\s*\[\s*\d+(?:,\s*\d+)*\s*\]\s*;?$',
        r'^\s*(?:\w+\s*=\s*)?IntegerDigits\s*\[\s*\d+\s*\]\s*;?$',
    ]
    
    # List to store matched programs and their indices
    matched_programs = []
    indexes = []

    # Iterate over each Mathematica code string
    for i, program in enumerate(mathematica_programs):
        # Preprocess the code string
        program = preprocess_code_mathematica(program)
        
        # Check if the program matches any of the patterns
        for pattern in patterns:
            if re.match(pattern, program, re.MULTILINE):
                matched_programs.append(program)
                indexes.append(i)
                break  # Move to the next program if a match is found
    
    return matched_programs, indexes
````

````python
# Find matched programs
matches, indexes = find_matched_programs_mathematica(test_answers)

# Print the matches and their indexes
for match, index in zip(matches, indexes):
    print(f"Matched program at index {index}:")
    print(match)
    print()  # Add an empty line for clarity
````

````python
mathe_df.loc[indexes, 'print_code'] = 1
mathe_df
````

````python
mathe_df.to_csv("automatic_mathematica.csv", index=False)
````
