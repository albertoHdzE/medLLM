# Analyzing Binary sequences from Zenil's paper

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
timeGPT_bin_df = pd.read_csv("bin_seq_last_digit_pred_timeGPT.csv")
chronos_bin_df = pd.read_csv("bin_seq_last_digit_pred_chronos.csv")
llama_bin_df = pd.read_csv("bin_seq_last_digit_pred_lag_llama.csv")

total_df = pd.concat(
    [timeGPT_bin_df, chronos_bin_df, llama_bin_df]
)
total_df
```

```python
# Calculate success percentage by model step by step
models = total_df['model'].unique()
success_rates = []

for model in models:
    # Get data for this model
    model_data = total_df[total_df['model'] == model]
    
    # Calculate success rate
    success_rate = (model_data['success'].sum() / len(model_data)) * 100
    print("model: ", model)
    print("sum: ", model_data['success'].sum())
    print("len: ", len(model_data))
    success_rates.append(success_rate)

# Add 'CTM/BDM' with a success rate of 100% at the beginning
models = np.insert(models, 0, 'ASI')
success_rates.insert(0, 100)

# Create bar plot using default color cycle with different colors for each bar
plt.figure(figsize=(10,6))
colors = plt.cm.tab10.colors  # Use the default color cycle
plt.bar(models, success_rates, color=colors[:len(models)])

# Increase font sizes
plt.title('Success Rate by Model - Simple climbers ', fontsize=24)
plt.xlabel('Model', fontsize=20)
plt.ylabel('Success Rate (%)', fontsize=20)
plt.xticks(fontsize=16, rotation=0)
plt.yticks(fontsize=16)

# Set y-axis limits from 0 to 100
plt.ylim(0, 100)

plt.tight_layout()
plt.show()

# Print success percentages as dataframe
success_df = pd.DataFrame({'Success Rate (%)': success_rates}, index=models)
print("\nSuccess Rates by Model:")
print(success_df)
```

```python
# Figure 1, top panel -- "Success Rate by Model - Simple climbers".
#
# Verification driver. The logic lives in superarc.timeseries; this cell shows
# the success rates before plotting so the bars can be read off the numbers.
# Regenerating here reproduces the committed plots/highResolution/figure01
# exactly (0.0000% pixel difference) -- checked by tests/test_timeseries.py.

from superarc.timeseries import BINARY_PANELS, binary_success_rates, plot_binary_panels

display(binary_success_rates(BINARY_PANELS["figure01"][1]).round(2))
```
